# Graphormer PCQM4Mv2 demo

Compute and plot $S_{\mathrm{sem}}$, $S_{\mathrm{str}}$, $J$, $D_{\mathrm{rel}}$, and $C_c(t,h,d)$ for the public checkpoint.

## Install

```python
import os
import subprocess
import sys
from pathlib import Path

REPO_URL = "https://github.com/joshgreenwa/Graph-Specialisation-and-Metrics.git"
REPO_DIR = Path("Graph-Specialisation-and-Metrics")

if Path("../src/graph_specialisation_metrics").is_dir():
    os.chdir("..")
elif not Path("src/graph_specialisation_metrics").is_dir():
    if not REPO_DIR.exists():
        subprocess.run(
            ["git", "clone", "--depth", "1", "--branch", "main",
             "--single-branch", REPO_URL, str(REPO_DIR)],
            check=True,
        )
    os.chdir(REPO_DIR)

subprocess.run([sys.executable, "-m", "pip", "install", "-q", "-e", ".[graphormer]"], check=True)
```

## Score

The first run downloads the public model and PCQM4Mv2 data. Set `FAST = False` to use the full configuration.

```python
FAST = True
OUTPUT_DIR = Path("results/graphormer_pcqm4mv2_demo")
command = [
    sys.executable, "-m", "graph_specialisation_metrics.cli", "score",
    "--config", "configs/graphormer_pcqm4mv2.yaml",
    "--output-dir", str(OUTPUT_DIR),
]
if FAST:
    command.append("--fast")
subprocess.run(command, check=True, input="y\n", text=True)
```

## Scores

```python
from graph_specialisation_metrics.plotting import load_scores, plot_scores

scores = load_scores(OUTPUT_DIR / "scores.npz")
plot_scores(scores)
```
